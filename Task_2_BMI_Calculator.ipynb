```python
# Task 2: BMI Calculator

## Description
This Python program calculates Body Mass Index (BMI)
using the user's weight and height.

## Features
- Calculates BMI
- Displays BMI category
- Validates user input
- Handles invalid values
- Allows repeated calculations



# Task 2 - BMI Calculator

print("===== BMI CALCULATOR =====")

while True:
    try:
        # User se input lena
        weight = float(input("Enter your weight (kg): "))
        height = float(input("Enter your height (meters): "))

        if weight <= 0 or height <= 0:
            print("❌ Error: Weight and height must be positive values.")
            continue

        # BMI calculation
        bmi = weight / (height ** 2)

        print("\n--------------------------")
        print(f"Your BMI is: {bmi:.2f}")

        if bmi < 18.5:
            print("Category: Underweight")
        elif bmi < 25:
            print("Category: Normal weight")
        elif bmi < 30:
            print("Category: Overweight")
        else:
            print("Category: Obese")

        print("--------------------------")

        again = input("\nDo you want to calculate again? (yes/no): ")

        if again.lower() != "yes":
            print("\nThank you for using BMI Calculator!")
            break

    except ValueError:
        print("❌ Error: Please enter numbers only.")
```
